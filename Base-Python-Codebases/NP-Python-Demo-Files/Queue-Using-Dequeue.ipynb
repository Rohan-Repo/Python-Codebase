### Queue using deque

```python
from collections import deque
import time
```

```python
queueList = deque([ 'Jessica', 'Harvey', 'Louis' ])

print( 'Original Queue : ', queueList )

# Enqueue -> Add to end
print( '\nEnqueue - Add to the Last using append()' )

queueList.append( 'Mike' )
print( 'After Person #1 : ', queueList )

queueList.append( 'Robert' )
print( 'After Person #2 : ', queueList )


# Dequeue -> Remove from front
print( '\nDequeue - Remove the First using popleft()' )

for i in range( len(queueList) ):

    print( '\nRound #', i+1 )
    time.sleep (1)

    print( 'Queue Before Removal : ', queueList )

    removedPerson = queueList.popleft()

    print( 'Removed Person : ', removedPerson )
    print( 'Queue After Removal : ', queueList )
```
