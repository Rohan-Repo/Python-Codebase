**Code to Update the Participant's Name based on NameList CSV File and the Graduation Date**

```python
#Create a certificate template with the name "Certificate.pdf"
#You should have CSV file of students name- Name it  "NameList.csv" .
####In your csv file , there should be only one column wth the header as "Name"
#Upload both the files to this script on left hand side where you see the box symbol.
#Execute this script

# Modules to print Name & Date
!pip install fitz
!pip install PyMuPDF

# Modules to flatten the PDF
!pip install fillpdf
from fillpdf import fillpdfs

# Module to read NameList CSV file
import pandas as pd

# Module to save files onto our Google Drive
import os.path
from google.colab import drive
drive.mount('/content/gdrive', force_remount=True)

import fitz

# Read Names from NameList CSV file and add it to a Python List
df=pd.read_csv("NameList.csv")
L=df['Name'].values.tolist()


pdf_files=[]

participantName = ''

for i in range(0,len(L)):

    participantName=L[i]

    # Update Graduation Date Below
    graduationDate = '2024-06-28'

    print( '\t For Particiant  : ', participantName, ' : ', graduationDate, 'len:', len(participantName) )


    # Normalized RGB color values
    darkslategray = (47/255, 79/255, 79/255)  #
    steelblue = (70/255, 130/255, 180/255)  # Normalized RGB
    teal = (0,128/255,128/255)

#     colors = {
#     "SteelBlue": (70/255, 130/255, 180/255),
#     "FireBrick": (178/255, 34/255, 34/255),
#     "MediumSeaGreen": (60/255, 179/255, 113/255),
#     "DarkOrange": (255/255, 140/255, 0),
#     "DeepSkyBlue": (0, 191/255, 255/255)
# }

    # Open the template Certificate PDF file
    pdf_document = fitz.open("Certificate.pdf")

    # Iterate through each page in the document
    for page_num in range(len(pdf_document)):
        page = pdf_document.load_page(page_num)
        # Update Partcipant Name
        # page.insert_text((275, 325), f"{participantName}", fontsize=28, color=(1,0,1))
        if len(participantName) >= 40:
          page.insert_text((275, 325), f"{participantName}", fontsize=22, fill=darkslategray,color=darkslategray)
        elif len(participantName) >= 30 and len(participantName) <= 39:
          page.insert_text((350, 325), f"{participantName}", fontsize=24, fill=darkslategray,color=darkslategray)
        elif len(participantName) >= 20 and len(participantName) <= 29:
          page.insert_text((400, 325), f"{participantName}", fontsize=26, fill=darkslategray,color=darkslategray)
        else:
          page.insert_text((450, 325), f"{participantName}", fontsize=28, fill=darkslategray,color=darkslategray)

        # Update Graduation Date
        # page.insert_text((480, 465), f"{graduationDate}", fontsize=20, color=(0,0.5,0.5))
        page.insert_text((480, 465), f"{graduationDate}", fontsize=20, fill=teal,color=teal)


    certPath="/content/gdrive/My Drive/Certificates"

    # If Certifictaes folder is not found create it
    if not os.path.isdir(certPath):
      os.makedirs(certPath)

    completeName = os.path.join(certPath,participantName+".pdf")

    print( '\t\t PDF filePath : ', completeName )

    # Save the modified PDF
    pdf_document.save( completeName )

    fillpdfs.flatten_pdf(completeName,completeName)
    pdf_files.append(completeName)

print('\n List of All PDF Files: ', pdf_files)
```
