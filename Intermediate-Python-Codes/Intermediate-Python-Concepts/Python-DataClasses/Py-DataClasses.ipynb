## Older way of Writing Classes

```python
class PersonCls:
    
    def __init__( self, name, age ):
        self.id = uuid.uuid4()
        self.name = name
        self.age = age

    def __repr__( self  ):
        return f" PersonCls ( '{self.id}' , '{self.name}', '{self.age}' )"

p1 = PersonCls( 'John Doe', 25 )
print( p1 )      

p2 = PersonCls( 'Jane Doe', 18 )
print( p2 )      
```

## For Classes created specifically for being Data-Focussed **DataClasses** are a concise way which provide a way to store data and provides several special methods like __init__(Constructor), __repr__(toString) and __eq__(compare) to reduce boilerplate code

### **@dataclass** decorator was introduced in Python 3.7 internally creates Constructor, toString() and equals() methods

```python
from dataclasses import dataclass, field

@dataclass
class PersonDataClass:
    # Set Default Values and Autogenerate UUID 
    id: uuid.UUID = field(default_factory=uuid.uuid4)
    name: str = ""
    age: int = 0

p3 = PersonDataClass( name='John Doe', age=25 )
print( p3 )      

p4 = PersonDataClass( name='Jane Doe', age=18 )
print( p4 )
```

### **Immutable** class where values should not change - **@dataclass(frozen=True)**

```python
@dataclass( frozen = True )
class FixedData():
    PI : float = 3.141592
    days : tuple = ( 'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun' )
    vowels : tuple = ( 'A', 'E', 'I', 'O', 'U' )

fd = FixedData()
print( 'PI : ', fd.PI )
print( 'Days : ', fd.days )
print( 'Vowels : ', fd.vowels )
print( 'Hash Value of Class as frozen = True generates a __hash__ by default : ', hash( fd ) ) 
print( fd ) 
```

### If we try to change a value it will throw a **FrozenInstanceError** : cannot assign to field 'PI'

```python
fd.PI = 1.5
```

### **Inheritance** - Don't need to send values to parent init class just inherit and datclass handles the initialization

```python
@dataclass
class PersonDataClass:
    # Set Default Values and Autogenerate UUID 
    id: uuid.UUID = field(default_factory=uuid.uuid4)
    name: str = ""
    age: int = 0

# Every Employee is a person so they inherit data from the Person class
@dataclass
class EmployeeDataClass(PersonDataClass):
    empId: uuid.UUID = field(default_factory=uuid.uuid4)
    dept: str = ""
    salary: float = 0

emp1 = EmployeeDataClass( name='John Doe', age=25, dept='IT', salary=100000)
print( emp1 )

emp2 = EmployeeDataClass( name='Jane Doe', age=18, dept='Sales', salary=50000)
print( emp2 )
```

### **Nested DataClass** - Class within Class

```python
@dataclass
class Address:
    street: str
    city: str
    zipCode : str
    
@dataclass
class PersonDataClass:
    # Set Default Values and Autogenerate UUID 
    id: uuid.UUID = field(default_factory=uuid.uuid4)
    name: str = ""
    age: int = 0
    address: Address = None

# Every Employee is a person so they inherit data from the Person class
@dataclass
class EmployeeDataClass(PersonDataClass):
    empId: uuid.UUID = field(default_factory=uuid.uuid4)
    dept: str = ""
    salary: float = 0

addr1 = Address( '123 Main St.', 'Toronto', 'M5J 1E6' )
emp1 = EmployeeDataClass( name='John Doe', age=25, address=addr1, dept='IT', salary=100000)
print( emp1 )

addr2 = Address( '456 Bloor St.', 'Toronto', 'M6P 1A5' )
emp2 = EmployeeDataClass( name='Jane Doe', age=18, address=addr2, dept='Sales', salary=50000)
print( emp2 )
print( 'Check out Individual Values:' )
print( '\t Emp Name : ', emp1.name, '\t', 'Emp Dept : ', emp1.dept, '\t', 'Emp Street : ', emp1.address.street ) 
```

### Convert **DataClass Object into JSON** Object

```python
import json
from dataclasses import asdict

emp1_dict = asdict( emp1 )
emp1_dict['id'] = str( emp1_dict['id'] )
emp1_dict['empId'] = str( emp1_dict['empId'] )
emp1_json = json.dumps( emp1_dict ) 
emp1_json
```

### Convert **DataClass Objects into JSON** Array

```python
empList = [ emp1, emp2 ]

# Convert each Employee dataclass instance to dict
emp_dict_list = [ asdict(emp) for emp in empList ]

# Convert UUID fields to string for each employee dict
for emp_dict in emp_dict_list:
    if 'id' in emp_dict:
        emp_dict['id'] = str(emp_dict['id'])
    if 'empId' in emp_dict:
        emp_dict['empId'] = str(emp_dict['empId'])

emp_json_arr = json.dumps( emp_dict_list, indent=2 )
print(emp_json_arr)
```

```python
@dataclass
class Address:
    street: str
    city: str
    zipCode : str
    
@dataclass
class PersonDataClass:
    id: str = "" 
    name: str = ""
    age: int = 0
    address: Address = None

# Every Employee is a person so they inherit data from the Person class
@dataclass
class EmployeeDataClass(PersonDataClass):
    empId: str = ""
    dept: str = ""
    salary: float = 0

addr1 = Address( '123 Main St.', 'Toronto', 'M5J 1E6' )
emp1 = EmployeeDataClass( id='p1', name='John Doe', age=25, address=addr1, empId='emp1', dept='IT', salary=100000)
print( emp1 )

addr2 = Address( '456 Bloor St.', 'Toronto', 'M6P 1A5' )
emp2 = EmployeeDataClass( id='p2', name='Jane Doe', age=18, address=addr2, empId='emp2', dept='Sales', salary=50000)
print( emp2 )

emp3 = EmployeeDataClass( id='p1', name='John Doe', age=25, address=addr1, empId='emp1', dept='IT', salary=100000)
print( emp3 )
print( 'By defailt __eq__ method' )
print( '\t emp1 == emp2', emp1 == emp2 )
print( '\t emp1 == emp3', emp1 == emp3 )
```

### Comparison methods __lt__(), __le__(), __gt__(), and __ge__()

```python
from dataclasses import dataclass, field

@dataclass(frozen=True, order=True)
class Employee:
    id: int = 0
    name: str = ""
    age: int = 0
    salary: float = 0.0

# Example usage
e1 = Employee(id=1, name="Alice", age=30, salary=70000)
e2 = Employee(id=2, name="Bob", age=35, salary=75000)
e3 = Employee(id=1, name="Alice", age=30, salary=70000)

# Equality (from __eq__)
print('e1 == e3 : ', e1 == e3)  # True, same values
print('e1 == e2 : ', e1 == e2)  # False

# Ordering (from __lt__, __le__, __gt__, __ge__)
print('e1 < e2 : ', e1 < e2)   # True, compares fields in order (id, name, age, salary)
print('e1 <= e3 : ', e1 <= e3)  # True

# Hashing (__hash__)
my_set = {e1, e2, e3}
print('because e1 and e3 are equal and hash equal : ', len(my_set))  # 2, because e1 and e3 are equal and hash equal
```
