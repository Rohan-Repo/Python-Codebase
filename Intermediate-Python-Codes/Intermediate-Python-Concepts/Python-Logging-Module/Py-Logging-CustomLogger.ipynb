## Import all the required Modules

```python
import logging
import os
import platform
import getpass
import socket
```

### Create a /logs directory if it does not exist

```python
# Create logs directory if it doesn't exist
log_dir = "logs"
os.makedirs(log_dir, exist_ok=True)
log_dir
```

***

### Create a Custom Logger

```python
# Explicitly create a Custom Logger
appLogger = logging.getLogger( "myApp" )

# Set the Logging Level to Debug
appLogger.setLevel( logging.INFO )

# Remove existing handlers if any (to avoid conflicts)
if appLogger.hasHandlers():
    appLogger.handlers.clear()

# Define full log file path inside logs directory
app_log_file_path = os.path.join(log_dir, "appLogger.log")

print( 'app_log_file_path : ', app_log_file_path )

# Create a File Handler
appHandler = logging.FileHandler( app_log_file_path, mode='a+' )

# Give the Logs a Format
appFormatter = logging.Formatter( 
        '%(asctime)s - %(name)s - %(levelname)s - %(message)s',
        datefmt='%Y-%m-%d %H:%M:%S'
)
appHandler.setFormatter( appFormatter ) 

# Add Handler to Logger
appLogger.addHandler( appHandler )
```

## Since default Level is set to INFO it will ignore the Debug Messages

### Get Information of the System generating the Log

```python
# Hostname
hostname = socket.gethostname()
appLogger.info("Hostname : " + hostname)

# Username
username = getpass.getuser()
appLogger.info("Username : " + username)

# Home directory
home_dir = os.path.expanduser("~")
appLogger.info("Home Directory : " + home_dir)

# Platform info
platform_info = platform.platform()
appLogger.info( "Platform Info : " + platform_info)

appLogger.info("uName : " + str(platform.uname()) )
```

```python
appLogger.debug( 'This is a App Debug Message' ) # Will be ignored as level is set to INFO
appLogger.info( 'This is a App Info Message' )
appLogger.warning( 'This is a App Warning Message' )
appLogger.error( 'This is a App Error Message' )
appLogger.critical( 'This is a App Crtical Message' )
```
