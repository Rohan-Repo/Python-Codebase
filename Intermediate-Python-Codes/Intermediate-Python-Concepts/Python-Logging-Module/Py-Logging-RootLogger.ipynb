```python
import logging
```

### Level 1 - Debug, Level 2 - Info, Level 3 - Warning (DEFAULT LEVEL - Anything this and above gets printed), Level 4 - Error, Level 5 - Critical, 

#### Since we have not setup our custom logger it uses the root logger by default

```python
logging.debug( 'This is a App Debug Message' ) 
logging.info( 'This is a App Info Message' )
logging.warning( 'This is a App Warning Message' )
logging.error( 'This is a App Error Message' )
logging.critical( 'This is a App Crtical Message' )
```

### Create a Custom Root Logger and Set Logging Level to Debug

```python
# Explicitly create a Root Logger
rootLogger = logging.getLogger()

# Set the Logging Level to Debug
rootLogger.setLevel( logging.DEBUG )
```

```python
rootLogger.debug( 'This is a App Debug Message' ) 
rootLogger.info( 'This is a App Info Message' )
rootLogger.warning( 'This is a App Warning Message' )
rootLogger.error( 'This is a App Error Message' )
rootLogger.critical( 'This is a App Crtical Message' )
```

### Create a /logs directory if it does not exist

```python
import os

# Create logs directory if it doesn't exist
log_dir = "logs"
os.makedirs(log_dir, exist_ok=True)
log_dir
```

```python
# Remove existing handlers if any (to avoid conflicts)
if rootLogger.hasHandlers():
    rootLogger.handlers.clear()

# Define full log file path inside logs directory
root_log_file_path = os.path.join(log_dir, "rootLogger.log")

print( 'root_log_file_path : ', root_log_file_path )

# Create a File Handler
rootHandler = logging.FileHandler( root_log_file_path, mode='a+' )

# Give the Logs a Format
rootFormatter = logging.Formatter( '%(asctime)s - %(name)s - %(levelname)s - %(message)s' )
rootHandler.setFormatter( rootFormatter ) 

# Add Handler to Logger
rootLogger.addHandler( rootHandler )
```

```python
rootLogger.debug( 'This is a App Debug Message' ) 
rootLogger.info( 'This is a App Info Message' )
rootLogger.warning( 'This is a App Warning Message' )
rootLogger.error( 'This is a App Error Message' )
rootLogger.critical( 'This is a App Crtical Message' )
```
