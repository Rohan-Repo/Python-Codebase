### Object Orientation - Object Comparison and Sorting

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def print_details(self):
        print("\t Person's Details:")
        print('\t\t Name = ', self.name)
        print('\t\t Age = ', self.age, 'years old')

    # Less Than <
    def __lt__(self, other):
        return self.age < other.age

    # Greater Than >
    def __gt__(self, other):
        return self.age > other.age

    # Equal To ==
    # So two different people with the same age would be considered equal so if we want two separte people we will compare both name and age:
    def __eq__(self, other):
        # return self.age == other.age # person1 == person2 returns True
        return self.name == other.name and self.age == other.age


person1 = Person('John Doe', 25)
person1.print_details()

person2 = Person('Jane Doe', 18)
# person2 = Person('Jane Doe', 25)
person2.print_details()

print("\n Age Comparisons:")
print("\t person1 < person2 :", person1 < person2) # Internally Calls : person1.__lt__(person2)
print("\t person1 > person2 :", person1 > person2) # Internally Calls : person1.__gt__(person2)
print("\t person1 == person2:", person1 == person2) # Internally Calls : person1.__eq__(person2)

print( '\n Sorting List of People based on age:' )
pplList = [
    person1,
    person2,
    Person("Chandler Bing", 30),
    Person("Joey Tribbiani", 40),
    Person("Monica Geller", 45),
    Person("Ross Geller", 35),
    Person("Phoebe Buffay", 15),
    Person("Rachel Greene", 21)
]

print( 'People Initially : ' )
for ppl in pplList:
    print( '\t', ppl.name, ' is ', ppl.age, ' years old' )
    
pplList.sort()

print( 'People After Sorting : ' )
for ppl in pplList:
    print( '\t', ppl.name, ' is ', ppl.age, ' years old' )
```

```python
class Person:

    # Class variable shared by all Person objects - Each department is assigned a rank that will be used for sorting.
    department_order = {
        "Manager": 1,
        "IT": 2,
        "Legal": 3,
        "Tech Support": 4
    }

    # Constructor method - Runs automatically whenever a new Person object is created.
    def __init__(self, name, age, department):
        self.name = name
        self.age = age
        self.department = department

    # Displays the details of a Person object.
    def print_details(self):
        print('\t', self.name, " | ", self.age, " | ", self.department)

    # Returns the age of the person.
    # This method will be used as the sort key when sorting people by age.
    def sort_by_age(self):
        return self.age

    # Returns the ranking value associated with the person's department.
    # This method will be used as the sort key when sorting people by department.
    def sort_by_department(self):
        return Person.department_order[self.department]


# Create a list containing multiple Person objects.
pplList = [
    Person("John Doe", 25, "IT"),
    Person("Jane Doe", 18, "Legal"),
    Person("Chandler Bing", 30, "Manager"),
    Person("Joey Tribbiani", 40, "Tech Support"),
    Person("Monica Geller", 45, "Manager"),
    Person("Ross Geller", 35, "IT"),
    Person("Phoebe Buffay", 15, "Legal"),
    Person("Rachel Greene", 21, "Tech Support")
]

# Display the original list before sorting.
print("People Initially:")
for person in pplList:
    person.print_details()


# Sort the list in ascending order of age.
# Python calls sort_by_age() for each Person object and uses the returned age value to determine the order.
print("\nPeople Sorted by Age Ascending:")

pplList.sort(key=Person.sort_by_age)

for person in pplList:
    person.print_details()


# Sort the list using the custom department ranking.
# reverse=True sorts from highest rank value to lowest:
# Tech Support (4) -> Legal (3) -> IT (2) -> Manager (1)
print("\nPeople Sorted by Department Descending:")

pplList.sort(
    key=Person.sort_by_department,
    reverse=True
)

for person in pplList:
    person.print_details()
```

### Sorting 
- .sort() 
  - changes the original pplList
  - only works on Lists
- .sorted() : 
  - creates and returns a new sorted list
  - pplList remains unchanged.
  - Works with many iterable objects 

```python
class Person:

    # Class variable containing the custom sorting order for each department.
    # The number represents the department's sorting priority.
    department_order = {
        "Manager": 1,
        "IT": 2,
        "Legal": 3,
        "Tech Support": 4
    }

    # Constructor method used to initialize each Person object.
    def __init__(self, name, age, department):
        self.name = name
        self.age = age
        self.department = department

    # Displays the person's name, age, and department.
    def print_details(self):
        print('\t', self.name, "|", self.age, "|", self.department)

    # Returns the person's age.
    # This method will be used by sorted() as the sorting key.
    def sort_by_age(self):
        return self.age

    # Returns the numerical ranking of the person's department.
    # This allows sorted() to sort departments according to our custom order.
    def sort_by_department(self):
        return Person.department_order[self.department]


# -------------------------------------------------
# CREATE LIST OF PEOPLE
# -------------------------------------------------

# Create a list containing multiple Person objects.
pplList = [
    Person("John Doe", 25, "IT"),
    Person("Jane Doe", 18, "Legal"),
    Person("Chandler Bing", 30, "Manager"),
    Person("Joey Tribbiani", 40, "Tech Support"),
    Person("Monica Geller", 45, "Manager"),
    Person("Ross Geller", 35, "IT"),
    Person("Phoebe Buffay", 15, "Legal"),
    Person("Rachel Greene", 21, "Tech Support")
]


# -------------------------------------------------
# DISPLAY ORIGINAL LIST
# -------------------------------------------------

# Display the list before applying any sorting.
print("People Initially:")

for person in pplList:
    person.print_details()


# -------------------------------------------------
# SORT BY AGE
# -------------------------------------------------

print("\nPeople Sorted by Age:")

# sorted() creates and returns a NEW sorted list.
# key=Person.sort_by_age tells Python to use the person's age
# as the value used for sorting.
sorted_by_age = sorted(
    pplList,
    key=Person.sort_by_age
)

# Display the newly sorted list.
for person in sorted_by_age:
    person.print_details()


# -------------------------------------------------
# SORT BY DEPARTMENT
# -------------------------------------------------

print("\nPeople Sorted by Department:")

# sorted() uses sort_by_department() as the sorting key.
# The method converts each department into its assigned number:
#
# Manager     -> 1
# IT          -> 2
# Legal       -> 3
# Tech Support -> 4
#
# Therefore, the list is sorted according to our custom
# department order rather than alphabetical order.
sorted_by_department = sorted(
    pplList,
    key=Person.sort_by_department
)

# Display the newly sorted list.
for person in sorted_by_department:
    person.print_details()


# -------------------------------------------------
# SORT BY DEPARTMENT - DESCENDING
# -------------------------------------------------

print("\nPeople Sorted by Department Descending:")

# reverse=True reverses the sorting order.
#
# Normal order:
# Manager -> IT -> Legal -> Tech Support
#
# Descending order:
# Tech Support -> Legal -> IT -> Manager
sorted_by_department_desc = sorted(
    pplList,
    key=Person.sort_by_department,
    reverse=True
)

# Display the newly sorted list.
for person in sorted_by_department_desc:
    person.print_details()
```

### With Lambda functions 

```python
class Person:

    # Custom sorting order for departments
    department_order = {
        "Manager": 1,
        "IT": 2,
        "Legal": 3,
        "Tech Support": 4
    }

    # Constructor
    def __init__(self, name, age, department):
        self.name = name
        self.age = age
        self.department = department

    # Display person details
    def print_details(self):
        print('\t', self.name, "|", self.age, "|", self.department)


# -------------------------------------------------
# CREATE LIST OF PEOPLE
# -------------------------------------------------

pplList = [
    Person("John Doe", 25, "IT"),
    Person("Jane Doe", 18, "Legal"),
    Person("Chandler Bing", 30, "Manager"),
    Person("Joey Tribbiani", 40, "Tech Support"),
    Person("Monica Geller", 45, "Manager"),
    Person("Ross Geller", 35, "IT"),
    Person("Phoebe Buffay", 15, "Legal"),
    Person("Rachel Greene", 21, "Tech Support")
]


# -------------------------------------------------
# DISPLAY ORIGINAL LIST
# -------------------------------------------------

print("People Initially:")

for person in pplList:
    person.print_details()


# -------------------------------------------------
# SORT BY AGE USING LAMBDA
# -------------------------------------------------

print("\nPeople Sorted by Age:")

# lambda person: person.age
# means: "Take each person and return their age."
# sorted() then uses that returned age as the sorting value.

# sort_by_age class method eliminated because of the Lambda function

sorted_by_age = sorted(
    pplList,
    key=lambda person: person.age
)

for person in sorted_by_age:
    person.print_details()


# -------------------------------------------------
# SORT BY DEPARTMENT USING LAMBDA
# -------------------------------------------------

print("\nPeople Sorted by Department:")

# lambda person: Person.department_order[person.department]

# For each person:
# 1. Get their department
# 2. Look up the department's ranking
# 3. Use that ranking for sorting
#
# Manager      -> 1
# IT           -> 2
# Legal        -> 3
# Tech Support -> 4

# sort_by_department class method eliminated because of the Lambda function
sorted_by_department = sorted(
    pplList,
    key=lambda person: Person.department_order[person.department]
)

for person in sorted_by_department:
    person.print_details()


# -------------------------------------------------
# SORT BY DEPARTMENT - DESCENDING
# -------------------------------------------------

print("\nPeople Sorted by Department Descending:")

# reverse=True reverses the custom department order.
#
# Tech Support -> 4
# Legal        -> 3
# IT           -> 2
# Manager      -> 1

sorted_by_department_desc = sorted(
    pplList,
    key=lambda person: Person.department_order[person.department],
    reverse=True
)

for person in sorted_by_department_desc:
    person.print_details()
```

| Feature                       | `.sort()`   | `sorted()`                | `lambda`                               |
| ----------------------------- | ----------- | ------------------------- | -------------------------------------- |
| What is it?                   | List method | Built-in Python function  | Anonymous function                     |
| Main purpose                  | Sort a list | Create a sorted result    | Define **what to sort by**             |
| Changes original list?        | ✅ Yes       | ❌ No                      | ❌ No                                   |
| Returns                       | `None`      | New sorted list           | A value/result                         |
| Works with                    | Lists       | Lists and other iterables | Used with functions such as `sorted()` |
| Can use `key=`?               | ✅ Yes       | ✅ Yes                     | Used **inside `key=`**                 |
| Can sort by custom attribute? | ✅           | ✅                         | ✅                                      |
| Can use `reverse=True`?       | ✅           | ✅                         | Not applicable                         |

| Sorting                | Lambda                                                      | What it returns       |
| ---------------------- | ----------------------------------------------------------- | --------------------- |
| **Age**                | `lambda person: person.age`                                 | Person's age          |
| **Name**               | `lambda person: person.name`                                | Person's name         |
| **Department**         | `lambda person: person.department`                          | Department name       |
| **Department Ranking** | `lambda person: Person.department_order[person.department]` | Custom ranking number |

```python
class Person:

    # Custom ranking assigned to each department
    department_order = {
        "Manager": 1,
        "IT": 2,
        "Legal": 3,
        "Tech Support": 4
    }

    # Constructor
    def __init__(self, name, age, department):
        self.name = name
        self.age = age
        self.department = department

    # Display person details
    def print_details(self):
        print('\t', self.name, "|", self.age, "|", self.department)


# -------------------------------------------------
# CREATE LIST OF PEOPLE
# -------------------------------------------------

pplList = [
    Person("John Doe", 25, "IT"),
    Person("Jane Doe", 18, "Legal"),
    Person("Chandler Bing", 30, "Manager"),
    Person("Joey Tribbiani", 40, "Tech Support"),
    Person("Monica Geller", 45, "Manager"),
    Person("Ross Geller", 35, "IT"),
    Person("Phoebe Buffay", 15, "Legal"),
    Person("Rachel Greene", 21, "Tech Support")
]


# -------------------------------------------------
# ORIGINAL LIST
# -------------------------------------------------

print("People Initially:")

for person in pplList:
    person.print_details()


# -------------------------------------------------
# 1. SORT BY AGE
# -------------------------------------------------

print("\nPeople Sorted by Age:")

sorted_by_age = sorted(
    pplList,
    key=lambda person: person.age
)

for person in sorted_by_age:
    person.print_details()


# -------------------------------------------------
# 2. SORT BY NAME
# -------------------------------------------------

print("\nPeople Sorted by Name:")

sorted_by_name = sorted(
    pplList,
    key=lambda person: person.name
)

for person in sorted_by_name:
    person.print_details()


# -------------------------------------------------
# 3. SORT BY DEPARTMENT ALPHABETICALLY
# -------------------------------------------------

print("\nPeople Sorted by Department:")

sorted_by_department = sorted(
    pplList,
    key=lambda person: person.department
)

for person in sorted_by_department:
    person.print_details()


# -------------------------------------------------
# 4. SORT BY CUSTOM DEPARTMENT RANKING
# -------------------------------------------------

print("\nPeople Sorted by Department Ranking:")

sorted_by_ranking = sorted(
    pplList,
    key=lambda person: Person.department_order[person.department]
)

for person in sorted_by_ranking:
    person.print_details()


# -------------------------------------------------
# 5. CUSTOM DEPARTMENT RANKING - DESCENDING
# -------------------------------------------------

print("\nPeople Sorted by Department Ranking Descending:")

sorted_by_ranking_desc = sorted(
    pplList,
    key=lambda person: Person.department_order[person.department],
    reverse=True
)

for person in sorted_by_ranking_desc:
    person.print_details()
```
