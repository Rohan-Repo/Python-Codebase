```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect("Product-DB/Product-Orders.db")

df_Products = pd.read_sql_query( 'SELECT * FROM Products', conn )
df_Customers = pd.read_sql_query( 'SELECT * FROM Customers', conn )
df_Orders = pd.read_sql_query( 'SELECT * FROM Orders', conn )
```

### INNER JOIN - Display Only Common Values - Intersection
- **Inner Join or Join** – This is similar to Intersection of two sets 
- Here, we get just the matching rows from both tables and eliminate others

```python
df_1_inner = pd.read_sql_query( 'SELECT o.orderDateTime, c.customerName, p.productName, p.productPrice, o.orderQuantity, p.productPrice * o.orderQuantity AS totalAmount FROM Customers c INNER JOIN Orders o ON o.customerID = c.customerID INNER JOIN Products p ON o.productID = p.productID', conn )
print( 'Inner Join Shape = ', df_1_inner.shape )
df_1_inner.tail()
```

### RIGHT JOIN - All Values from Right Table and only Matching Values from the Left Table
- **Right Join or Right Outer Join** – This is similar to Set Difference.
- Here we get the matching rows in addition to the non-matching rows from the Right table.
- Get the matching rows + non-matching rows from the Right table 

```python
df_1_right = pd.read_sql_query( 'SELECT o.orderDateTime, c.customerName, p.productName, p.productPrice, o.orderQuantity, p.productPrice * o.orderQuantity AS totalAmount FROM Customers c RIGHT JOIN Orders o ON o.customerID = c.customerID RIGHT JOIN Products p ON o.productID = p.productID', conn )
print( 'Right Join Shape = ', df_1_right.shape )
df_1_right.tail()
```

### LEFT JOIN - All Values from Left Table and only Matching Values from the Right Table
- **Left Join or Left Outer Join** – This is similar to Set Difference. 
- Here we will get the matching rows in addition to the non-matching rows from the Left table
- Get the matching rows + non-matching rows from the Left table

```python
df_1_left = pd.read_sql_query( 'SELECT o.orderDateTime, c.customerName, p.productName, p.productPrice, o.orderQuantity, p.productPrice * o.orderQuantity AS totalAmount FROM Customers c LEFT JOIN Orders o ON o.customerID = c.customerID LEFT JOIN Products p ON o.productID = p.productID', conn )
print( 'Left Join Shape = ', df_1_left.shape )
df_1_left.tail(10)
```

### Cross Join – Here we get the cartesian product
- **Cartesian Product = number of rows of table1 * number of rows of table2**
- Since Cross Joins give us the Cartesian Product, we don’t need to specify a condition

```python
df_1_cross = pd.read_sql_query( 'SELECT c.customerName, p.productName FROM Customers c CROSS JOIN Products p', conn )
print( 'Products Shape = ', df_Products.shape )
print( 'Customers Shape = ', df_Customers.shape )
print( 'Cross Join Shape = ', df_1_cross.shape )
print( '\n Cross Join Head = \n', df_1_cross.head() )
print( '\n Cross Join Tail = \n', df_1_cross.tail() )
```

### Merge DataFrames using Left, Right Outer and Inner : .merge()
- **Similar to SQL JOINS**
- Merge DataFrame1 and DataFrame2 and Pass as an Argument with DataFrame3

#### Inner Join

```python
df_2_inner = pd.merge( 
    pd.merge(df_Customers, df_Orders, on='customerID', how='inner'),
    df_Products,
    on='productID', how='inner')
print( 'Inner Join Shape = ', df_2_inner.shape )
df_2_inner.tail()
```

### Outer Join:
- Return Matched and Unmatched Values from either or both the Tables

```python
df_2_outer = pd.merge( 
    pd.merge(df_Customers, df_Orders, on='customerID', how='outer'),
    df_Products,
    on='productID', how='outer')
print( 'Outer Join Shape = ', df_2_outer.shape )
df_2_outer.tail(10)
```

#### Left Join

```python
df_2_left = pd.merge( 
    pd.merge(df_Customers, df_Orders, on='customerID', how='left'),
    df_Products,
    on='productID', how='left')
print( 'Left Join Shape = ', df_2_left.shape )
df_2_left.tail(10)
```

#### Right Join

```python
df_2_right = pd.merge( 
    pd.merge(df_Customers, df_Orders, on='customerID', how='right'),
    df_Products,
    on='productID', how='right')
print( 'Right Join Shape = ', df_2_right.shape )
df_2_right.tail()
```

### Concatenate DataFrames using Left, Right Outer and Inner : .concat()
- **Add Rows** *(one below the Other)* or **Columns** *(Side by Side)* and Add **NaN for missing** items
- Specify **join** type
- Also Specify the **Axis**, **axis=0** means **concatenate** along the **rows** and **axis=1** means **concatenate** along the **columns**
- **ignore_index=True**: This resets the index after concatenation, creating a new default integer index rather than retaining the original indices.

#### Inner Concat

```python
df_3_inner = pd.concat( [df_Customers, df_Orders, df_Products], axis=1, join='inner', ignore_index=True )
print( 'Inner Concat Shape = ', df_3_inner.shape )
df_3_inner.tail()
```

#### Outer Concat

```python
df_3_outer = pd.concat( [df_Customers, df_Orders, df_Products], axis=1, join='outer', ignore_index=True )
print( 'Outer Concat Shape = ', df_3_outer.shape )
df_3_outer.tail()
```

### Right or Left Concat Not Allowed : ValueError: Only can inner (intersect) or outer (union) join the other axis

### join() is used to combine DataFrames based on their indices.
- **join() merges two or more tables horizontally (side by side)**
- **Both DataFrames must have the same or compatible indices.**

```python
# Customers DataFrame
df_cust = pd.DataFrame({
    'Name': ['Chandler', 'Monica', 'Ross'],
    'City': ['New York', 'Seattle', 'Long Island']
}, index=[1, 2, 3])

df_cust
```

```python
# Orders DataFrame
df_ord = pd.DataFrame({
    'OrderID': [101, 102, 103],
    'ProductID': [11, 12, 13],
    'Quantity': [2, 1, 4]
}, index=[1, 2, 3])
df_ord
```

```python
# Products DataFrame
df_prod = pd.DataFrame({
    'ProductName': ['iPhone', 'Laptop', 'Smartwatch'],
    'Price': [999, 1899, 299]
}, index=[11, 12, 13])
df_prod
```

#### Same Indices for both the DataFrames

```python
df_cust_ord = df_cust.join( df_ord )
df_cust_ord
```

#### Join products using ProductID as a key (need to reset index alignment)

```python
df_join = df_cust_ord.join( df_prod, on='ProductID' )
df_join
```

### join() merges side by side based on index by default.

#### When joining with Products, we use on=*'ProductID'* since its index *(11,12,13)* matches the values in the *ProductID* column of Order DataFrame.
