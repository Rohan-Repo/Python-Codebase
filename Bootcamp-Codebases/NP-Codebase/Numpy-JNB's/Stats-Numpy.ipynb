```python
import numpy as np
```

### Statistical Operations in Numpy

```python
vector = np.array([20, 49, 33, 23, 54, 53, 23, 18, 27, 38]) 
print( '\n Vector : ', vector )

print( '\n Basic Statistics:' )
print( f" \t Sum = {np.sum(vector)} " ) # Total
print( f" \t Mean = {np.mean(vector)} " ) #Average

# Median: the middle value in a dataset when the numbers are arranged in 
#  ascending or descending order. If there are two middle values 
# (even count), the median is their average.  
print( f" \t Median = {np.median(vector)} " ) 

# Standard deviation: measures how spread out values are from the mean; 
# a higher value means more variability, while a lower value 
# means data is closer to the average.  
print( f" \t Standard Deviation = { round(np.std(vector),4) } " )

# Minimum and max from Vector
print( f" \t Min = {np.min(vector)} " )
print( f" \t Max = {np.max(vector)} " )
print( f" \t Index of Min = {np.argmin(vector)} " )
print( f" \t Index of Max = {np.argmax(vector)} " )

# Range of data (max - min)
print( f" \t Range (Max-Min) : {np.max(vector)} - {np.min(vector)} = {np.max(vector) - np.min(vector)}" )
```
