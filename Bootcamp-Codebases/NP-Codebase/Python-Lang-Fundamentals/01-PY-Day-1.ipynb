### Basic Calculator - Arithmetic Operations
#### Take user Input in String Format with input(msg)

```python
num1 = input( 'Enter Number 1: ' )
num2 = input( 'Enter Number 2: ' )
```

#### Since I/P is String this is String Concatenation not Addition

```python
print( 'Not Addition of Numbers but Concatenation = ', num1+num2) 
print( 'Value : ', num1, '\t Type : ', type(num1) )
print( 'Value : ', num2, '\t Type : ', type(num2) )
```

#### let's typecast[Convert] to Integer

```python
num1 = int( num1 )
num2 = int( num2 )
print( 'Value : ', num1, '\t Type : ', type(num1) )
print( 'Value : ', num2, '\t Type : ', type(num2) )
```

### Arith Operations

```python
print( 'Basic Calculator:' )
print( '\t Addition = ', num1 + num2 )
print( '\t Subtraction = ', num1 - num2 )
print( '\t Multiplication = ', num1 * num2 )
print( '\t Division = ', num1 / num2 )
print( '\t Division Rounded to 2 Decimal Places = ', round( (num1/num2), 2 ) )
print( '\t Remainder = ', num1 % num2 )
print( '\t Power = ', num1 ** num2 )
# the floor division // rounds the result down to the nearest whole number 
print( '\t Floor Division = ', num1 // num2 )
```

#### Customize Code and O/P as per requirement

```python
print( 'Functions with parameters:' )
def greetUser( msg ):
    print( '\t Parameter passed: ', msg )

greetUser( 'Bonjour!' )
greetUser( "¡Buenos días!" )
```

### Conditional: Make decision based on Condition

```python
print('\n Conditional Example:')
age = int( input('Enter Age : ' ) )

if age >= 18:
    print( "Person with age : ", age, " is Allowed to Vote" )
elif age < 0:
    print( "Person with age : ", age, ' is Not Even born yet!' )
else:
    print( "Person with age : ", age, " is Not Allowed to Vote " )
```
