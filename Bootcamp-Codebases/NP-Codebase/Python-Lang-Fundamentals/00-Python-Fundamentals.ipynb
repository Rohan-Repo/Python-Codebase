### Loops - Perform the same task over and over again

```python
dayList = [ 'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat' ]
print( 'Original  List : ',  dayList ) 
print( 'Built-In Function:' )
print( '\t Size of List Before : ', len(dayList) ) # Built-In Function Eg. 1 
dayList.append( 'Sun' ) # Built-In Function Eg. 2 
print( 'Final  List  : ',  dayList ) 
print( '\t Size of List After: ', len(dayList) )
```

#### While Loop → Condition-controlled loop (runs while a condition is True)
- Eg. ATM asking for PIN repeatedly until the correct PIN is entered.

```python
print('While Loop Example:' )
num = 0
while num < len(dayList):
    print( dayList[num], end=',' )
    num = num + 1
```

#### For Loop → Counter-controlled / sequence-controlled loop (iterates a fixed number of times or through a sequence)
- Eg. A data analyst generating monthly sales reports for all 12 months of the year.

```python
print('\n For Loop Example #2:')
for i in range( 7 ):
    print( i, ' : ', dayList[i], end='\t' )
```

#### forEach loop → Object/collection-controlled loop (iterates through elements of a collection like a list or array)
- Eg. Spotify going through each song in a playlist one by one.

```python
print('\n ForEach Loop Example :')
for day in dayList:
    print( day, end=' - ')
```

#### Nested Loop → Loop-inside-a-loop (one loop runs completely for each iteration of another loop)
- Eg. A data analyst reviewing sales data for every month and, within each month, checking every product category.

```python
print('\n Nested Loop Example :')
for i in range(1,3):
    for j in range(1,3):
        print( 'i : ',i,'j : ', j, end='\t')
    print('\n')
```

#### range() Function → Sequence generator 
- creates a sequence of numbers, commonly used with for loops
- **range(start, stop, step)**

```python
print( 'Odd Numbers : ', list( range( 1,10, 2 ) ) )
print( 'Even Numbers : ', list( range( 0,10, 2 ) ) )
```

### String Methods - String = Sequence of Characters
- Useful for textual data usage and manipulation

```python
data = '1,JohnDoe,sales,Toronto,Canada,$100000'

print( 'Normal-Print: ', data, '\t Type : ', type(data) )
print( 'String-Length : ', len(data) )
print( 'Upper Case : ', data.upper() )
print( 'Lower Case : ', data.lower() )
print( 'Swap Case : ', data.swapcase() )
print( 'Position of J : ', data.find('J') ) # first-occurence if found
print( 'Position of Z : ', data.find('Z') ) # otherwise returns -1

# Default Separator = Space
dataSplit = data.split()
print( 'Space: ', dataSplit )

# Our Separator = Comma
dataSplit = data.split(',')
print( 'Comma: ', dataSplit )

print( 'Capitalize : ', dataSplit[2].capitalize() )
print( 'Join via Delimiter : ', '-'.join(dataSplit) )
```

### Exception Handling → Error-handling mechanism (prevents a program from crashing when an error occurs)
- Eg. An ATM displays "Invalid PIN, please try again" instead of suddenly shutting down when the wrong PIN is entered.
- If something goes wrong, handle the problem and continue gracefully.

```python
try:
    # Code that can raise an Error
    # div = 10/0
    div = 10/2
except ZeroDivisionError:
    # The except block handles the error if it occurs.
    print( 'Division by Zero not allowed!' )
    # Multiple except blocks can handle different types of exceptions.
else:
    # The else block runs only if no exception occurs.
    print( 'Division Successful : ', div )
finally:
    # The finally block always executes whether an exception occurs or not.
    print( 'Code finished!' )
```

#### Class Example:
- **OOP** - Model real-world objects into Code
- `__init__` : A special constructor method in Python that runs when an object is created.
- **self** : Refers to the current object/instance of the class and is used to access its variables and methods.

```python
class Person():
    

    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def print_details( self ):
        print("Person's Details:")
        print( '\t Name = ', self.name )
        print( '\t Age = ', self.age ,' years old' )

person1 = Person( 'John Doe', 25 )
person1.print_details()

person2 = Person( 'Jane Doe', 18 )
person2.print_details()
```

### DateTime Formatting → Converting dates and times into a readable format

```python
# Date and Time Formatting

import datetime
date_now = datetime.datetime.now()

print( 'Default DateTime Format : ', date_now )
# Weekday, Day-Month(Str)-Year, Hour:Minutes:Seconds
print( 'Date Formatted : {:%A, %d-%B-%Y %I:%M:%S}'.format(date_now) )

import time 

time_now = time.asctime()
print( 'Default Time Format : ', time_now )

# Weekday, Year/Month(Num)/Day, Hour:Minutes:Seconds AM/PM Zone
time_now = time.strftime('%A, %Y/%m/%d,  %I:%M:%S %p %Z')
print( 'Time Formatted : {} \n'.format(time_now) )
```

### Match-Case (Python 3.10+) → Multi-option decision structure (checks one value against many possible choices)
- Similar to Switch case available in C++ or Java
- Eg. Airport self-check-in kiosk showing different information depending on the airport code entered.

```python
# Read user input
code = input("Enter Airport Code: ").upper()

match code:

    case "YYZ":
        print("Toronto :", code, "- Toronto Pearson Airport")

    case "YVR":
        print("Vancouver :", code, "- Vancouver International Airport")

    case "YUL":
        print("Montreal :", code, "- Montréal-Pierre Elliott Trudeau International Airport")

    case "YHZ":
        print("Halifax :", code, "- Halifax Stanfield International Airport")

    case _:
        print(code, "- Incorrect Airport Code")
```

### Hashing Example:
- Hashing is the process of converting data into a fixed-size value called a hash using a hash function.
- Making a small change like Toronto to toronto changes the Hash value drastically
- Used for secure password storage and data verification in cybersecurity.
- Used in dictionaries/databases for very fast data searching and retrieval.
- Examples of Algorithms : MD5, SHA-1, SHA-256

```python
import hashlib;

city = input( 'Enter your favourite City : ' )

print( 'String: ', city )

md5Val = hashlib.md5( city.encode() )

print( ' The hexadecimal equivalent of ', city ,' in MD5 is : = ', md5Val.hexdigest() )

sha1Val = hashlib.sha1( city.encode() )

print( ' The hexadecimal equivalent of ', city ,' in SHA1 is : = ', sha1Val.hexdigest() )
```
