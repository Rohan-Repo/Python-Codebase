### Create Random List of Numbers

```python
import random
```

```python
numList = []

cnt = int( input( 'Enter Count : ' ) ) 
start = int( input( 'Enter Start : ' ) ) 
end = int( input( 'Enter End : ' ) ) 

for i in range( cnt ):
    numList.append( random.randrange( start, end ) )

print( '\n Numbers List : ', numList )
```

```python
print( '\t Numbers Initially : ', numList )
numList.reverse()
print( '\t Numbers Reversed : ', numList )
numList.sort()
print( '\t Numbers Sort Ascending : ', numList )
numList.sort(reverse=True)
print( '\t Numbers Sort Descending : ', numList )

print( '\t Minimum out of All Numbers : ', min(numList) )
print( '\t Maximum out of All Numbers : ', max(numList) )
print( '\t Sum of All Numbers : ', sum(numList) )
print( '\t Count of List Items  : ', len(numList) )
print( '\t Average of All Numbers : ', sum(numList)/len(numList) )
```
