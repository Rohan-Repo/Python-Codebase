```python
# ==========================================
# Web Scraping quotes.toscrape.com
# Using Requests + BeautifulSoup
# Scrapes:
# - Quote Text
# - Author
# - Tags
# Saves to Pandas DataFrame + CSV
# ==========================================
```

### Import Required Libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

***

```python
# Website URL
url = "https://quotes.toscrape.com/"

# Send GET request
response = requests.get(url)

# Check if website loaded successfully
if response.status_code == 200:
    
    # Parse HTML
    soup = BeautifulSoup(response.text, "html.parser")

    # print('HTML:', soup)
    
    # Find all quote blocks
    quotes = soup.find_all("div", class_="quote")
    
    # Store scraped data
    quote_data = []

    print("\nQuotes Scraped:\n")

    for quote in quotes:

        # print( 'Q: ', quote )
        
        # Quote Text
        quote_text = quote.find("span", class_="text").get_text(strip=True)
        
        # Author
        author = quote.find("small", class_="author").get_text(strip=True)
        
        # Tags (multiple possible)
        tags = [tag.get_text(strip=True) for tag in quote.find_all("a", class_="tag")]
        
        # Convert tags list to comma-separated string
        tags_text = ", ".join(tags)
        
        # Save to list
        quote_data.append({
            "Quote Text": quote_text,
            "Author": author,
            "Tags": tags_text
        })
        
        # Print each quote
        print(f"Quote: {quote_text}")
        print(f"Author: {author}")
        print(f"Tags: {tags_text}")
        print("-" * 60)

    # ==============================
    # SAVE TO PANDAS DATAFRAME
    # ==============================
    df = pd.DataFrame(quote_data)

    # Save to CSV
    df.to_csv("quotes_scraped.csv", index=False)

    print("\nData saved to quotes_scraped.csv")

else:
    print("Failed to access website. Status Code:", response.status_code)
```
