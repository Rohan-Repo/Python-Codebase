```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect("ProductDB/Product.db")

df_Products = pd.read_sql_query( 'SELECT * FROM Products', conn )
df_Orders = pd.read_sql_query( 'SELECT * FROM Orders', conn )
```

### INNER JOIN - Display Only Common Values - Intersection

```python
df_1_inner = pd.read_sql_query( 'SELECT productName, productType, productPrice, orderQuantity, productPrice * orderQuantity AS totalAmount FROM Products INNER JOIN Orders ON Products.productID = Orders.productID', conn )
print( 'Inner Join Shape = ', df_1_inner.shape )
df_1_inner
```

### RIGHT JOIN - All Values from Right Table and only Matching Values from the Left Table

```python
df_1_right = pd.read_sql_query( 'SELECT productName, productType, productPrice, orderQuantity, productPrice * orderQuantity AS totalAmount FROM Products RIGHT JOIN Orders ON Products.productID = Orders.productID', conn )
print( 'Right Join Shape = ', df_1_right.shape )
df_1_right.tail()
```

### LEFT JOIN - All Values from Left Table and only Matching Values from the Right Table

```python
df_1_left = pd.read_sql_query( 'SELECT productName, productType, productPrice, orderQuantity, productPrice * orderQuantity AS totalAmount FROM Products LEFT JOIN Orders ON Products.productID = Orders.productID', conn )
print( 'Left Join Shape = ', df_1_left.shape )
df_1_left.tail(10)
```

### Cross Join – Here we get the cartesian product
- **Cartesian Product = number of rows of table1 * number of rows of table2**

```python
df_1_cross = pd.read_sql_query( 'SELECT * FROM Products CROSS JOIN Orders', conn )
print( 'Products Shape = ', df_Products.shape )
print( 'Orders Shape = ', df_Orders.shape )
print( 'Cross Join Shape = ', df_1_cross.shape )
df_1_cross
```

### Merge DataFrames using Left, Right Outer and Inner : .merge()
- **Similar to SQL JOINS**
- Merge DataFrame1 and DataFrame2 and Pass as an Argument with DataFrame3

#### Inner Join

```python
df_2_inner = pd.merge( df_Orders, df_Products, on='productID', how='inner')
print( 'Inner Join Shape = ', df_2_inner.shape )
df_2_inner.tail()
```

### Outer Join:
- Return Matched and Unmatched Values from either or both the Tables

```python
df_2_outer = pd.merge( df_Orders, df_Products, on='productID', how='outer')
print( 'Outer Join Shape = ', df_2_outer.shape )
df_2_outer
```

#### Left Join

```python
df_2_left = pd.merge( df_Orders, df_Products, on='productID', how='left')
print( 'Left Join Shape = ', df_2_left.shape )
df_2_left
```

#### Right Join

```python
df_2_right = pd.merge( df_Orders, df_Products, on='productID', how='right')
print( 'Right Join Shape = ', df_2_right.shape )
df_2_right.tail()
```

### join() is used to combine DataFrames based on their indices.
- **join() merges two or more tables horizontally (side by side)**
- **Both DataFrames must have the same or compatible indices.**

#### Let's Set the Index to Common Column ProductID for both the dataFrames

```python
df_Products = df_Products.set_index('productID')
df_Products
```

```python
df_Orders = df_Orders.set_index('productID')
df_Orders
```

```python
df_join = df_Products.join( df_Orders )
df_join
```
