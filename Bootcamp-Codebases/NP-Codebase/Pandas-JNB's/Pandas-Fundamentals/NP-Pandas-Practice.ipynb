### Add or Remove Columns

```python
print( 'Before : \n ', df_2[['Item','PricePerItem']].sort_values('PricePerItem').head() )

# Increase the Value of all Price by 20%
df_2['PricePerItem-2'] = df_2['PricePerItem'] + (df_2['PricePerItem'] * 0.2)
print( '\n After : \n ', df_2[['Item','PricePerItem-2']].sort_values('PricePerItem-2').head() )

# Recalculate TotalPrice
df_2['TotalPrice'] = df_2['PricePerItem-2'] * df_2['Quantity']

df_2[['Item','PricePerItem-2','Quantity','TotalPrice']].sort_values('TotalPrice',ascending=False).head()
```

#### Drop Columns

```python
df_2.drop(columns=['PricePerItem'])

# If we check we don't get any Change cause we didn't actually delete it
print( 'Not Deleteled : ', df_2.head() )

# To delete it use inplace
df_2.drop(columns=['PricePerItem'], inplace=True)
df_2.head()
```

#### Combine DateTime Columns together

```python
# pd.to_datetime - To Ensure the DateTime DataType for our DateTime instead of it being a String
df_2['DateTime'] = pd.to_datetime((df_2['Date'] + ' ' + df_2['Time']), format='%Y-%m-%d %H:%M:%S')

df_2.drop(columns=['Date','Time'],inplace=True)

df_2.head()
```

#### Rename and Reorder Column

```python
df_2 = df_2.rename(columns={'PricePerItem-2':'PricePerItem'} )
# Or we can do df_2.rename(columns={'PricePerItem-2':'PricePerItem'}, inplace=True )

columnOrder = ['TransactionID', 'DateTime', 'Item', 'PricePerItem', 'Quantity', 'TotalPrice',
       'PaymentMethod']
df_2 = df_2[columnOrder]

df_2.head()
```

#### Create a Copy of our DataFrame

```python
# If we do Assignment the new var just points to the original dataFrame
df_3 = df_2
print( 'MemLoc #1 : ', id( df_2 ), ' \t MemLoc #2 : ', id( df_3 ) )

# But if we want a separate copy we use copy()
df_copy = df_2.copy()
print( 'MemLoc #3 : ', id( df_2 ), ' \t MemLoc #4 : ', id( df_copy ) )
```

##### Drop the NAN's from Item and Payment column

```python
df_2.dropna(subset=['Item','PaymentMethod'],inplace=True)

df_2
```

##### Fill in Numeric Columns Item with Price as per ItemType, Quantity with the mode Value and re-calculate TotalPrice

```python
# Fill in PricePerItem with Price as per ItemType
df_2.loc[df_2['Item'] == 'Latte', 'PricePerItem'] = 4.80
df_2.loc[df_2['Item'] == 'Tea', 'PricePerItem'] = 2.70
df_2.loc[df_2['Item'] == 'Mocha', 'PricePerItem'] = 5.10

# Quantity with the mode Value
df_2['Quantity'] = df_2['Quantity'].fillna( df_2['Quantity'].mode()[0] )

# re-calculate TotalPrice
df_2['TotalPrice'] = df_2['Quantity'] * df_2['PricePerItem']
df_2
```

##### Grouby Cost As per ItemType - Highest Sold Items and Max Revenue PaymentMethod

```python
# Get Count GroupBy
print( 'Group By Item Type and Sum of TotalPrice:', df_2.groupby('Item')['TotalPrice'].sum().sort_values(ascending=False) )

print( '\n Group By PaymentMethod and Sum of TotalPrice:', df_2.groupby('PaymentMethod')['TotalPrice'].sum().sort_values(ascending=False) )
```
