#### Import Required Libraries

```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect("EmpDB/EmpDB.db")

df = pd.read_sql_query( 'SELECT * FROM Employee', conn )
df
```

### Display Pandas DataFrame as a HTML Document

##### Normal Display without Styling

```python
df.to_html( 'data1.html', index=False, border=1 )
```

##### CSS Styling Added

```python
# CSS styling for the HTML table and webpage
css = """
<style>
  body { background: lightgoldenrodyellow; margin: 0; padding: 20px; font-family: Arial, sans-serif; }
  .s2 { width: 100%; border-collapse: separate; font-size: 14px; }
  .s2 th { background: #0C447C; color: #E6F1FB; padding: 9px 14px; text-align: left; }
  .s2 td { background: #E6F1FB; padding: 8px 14px; color: #042C53; }
</style>
"""

# index=False removes row numbers and apply CSS styling specified above
html = df.to_html( index=False, border=0, classes='s2' )

# Create and write the styled HTML output into a file
with open('data2.html', 'w') as f:
    f.write(css + html)

# Close the database connection
conn.close()
```
